```python
import numpy as np
import os
import matplotlib.pyplot as plt
%config InlineBackend.figure_format = 'svg'
```

```python
ui = -0.1
Ti = 5
mi = 1836
vti = np.sqrt(Ti/mi)
```

```python
def fi(v):
    return 1/np.sqrt(2*np.pi)/vti*np.exp(-0.5*(v-ui)**2/vti**2)
```

```python
vh = 0.001
vs1 = -0.0743
vs2 = 0.0398
vs3 = 0.0

df1 = (fi(vs1+vh) - fi(vs1-vh))/(2*vh)
df2 = (fi(vs2+vh) - fi(vs2-vh))/(2*vh)
df3 = (fi(vs3+vh) - fi(vs3-vh))/(2*vh)
```

```python
print("Slope at vs1:", df1) 
print("Slope at vs2:", df2)
print("Slope at vs3:", df3)
```

```python
v = np.linspace(-0.15, 0.1, 200)
fm = fi(v)
plt.plot(v, fm, label='fi(v)')
plt.axvline(vs1, color='r', linestyle='--', label='vs1')
plt.axvline(vs2, color='g', linestyle='--', label='vs2')
plt.axvline(vs3, color='b', linestyle='--', label='vs3')
plt.xlabel('v')
plt.ylabel('fi(v)')
plt.legend()
plt.show()
```
