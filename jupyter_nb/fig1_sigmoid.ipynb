```python
import numpy as np
import os
import matplotlib.pyplot as plt
import matplotlib.colors as colors
import matplotlib.ticker as ticker
%config InlineBackend.figure_format = 'svg'
```

```python
plt.rcParams.update({'font.size': 16})
plt.rcParams.update({'lines.linewidth': 2})
```

```python
t = np.linspace(0, 15, num = 600)
t_start = 3.5
t_end = 11.5
tau = 0.1
```

```python
def S(t):
    return 1/(1+np.exp(-t))
sigma = S((t-t_start)/tau) + S(-(t-t_end)/tau) - 1
```

```python
fig = plt.figure(figsize=(6, 3))
plt.plot(t, sigma, color='tab:blue')
plt.axvline(t_start, color='tab:gray', linestyle='--',alpha=0.8)
plt.axvline(t_end, color='tab:gray', linestyle='--',alpha=0.8)
plt.xlabel('t')
plt.xticks([t_start,t_end], [r'$t_{start}$', r'$t_{end}$'])
plt.yticks([0,0.5, 1], ['0','', '1'])
plt.grid()
plt.title(r'$S[(t-t_{start})/\tau]+S[-(t-t_{end})/\tau] - 1$', fontsize=14)
# plt.savefig('fig1_sigmoid.pdf', bbox_inches='tight')
plt.show()
```
